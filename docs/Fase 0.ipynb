### **1. Sintaxe Fundamental da Linguagem Kotlin:**
<font size=3>

Nesta etapa, iremos aprender a declarar dados, tomar decisões e criar funções com a sintaxe moderna e enxuta do Kotlin.

* **Declaração de Variáveis Imutáveis e Mutáveis:**
    * `val`: Somente leitura (equivalente a constantes). **Boa prática:** Usar `val` por padrão.
    * `var`: Variável cujo valor pode ser sobrescrito.
    <br>

* **Inferência de Tipos:** O Kotlin identifica automaticamente tipos básicos como `Int`, `Double`, `Boolean` e `String`.
    * **Controle de Fluxo com Expressões:**
    * No Kotlin, estruturas como `if/else` e `when` retornam valores diretamente.
    * O `when` é a estrutura usada para checar condições múltiplas e faixas de valores.
    <br>

* **Funções e Declarações Simplificadas:**
    * Funções de corpo de expressão (linha única).
    * Parâmetros nomeados e com valores padrão (*default arguments*).

````kotlin
val groupName = "GTA"
var currentRssiReading = -65

// Estrutura 'when' como expressão para avaliar a qualidade do sinal Wi-Fi
val signalQuality = when (currentRssiReading){in -50..0 -> "Excelente"
                                              in -70..-51 -> "Bom"  
                                              in -85..-71 -> "Fraco"
                                              else -> "Sem Sinal"}

fun generateStatus(ssid: String = "WiFi_UFRN", quality: String): String =
    "Conectado à rede $ssid | Qualidade: $quality"

println(generateStatus(quality = signalQuality))
````

### **2. Sistema de Tipos e *Null Safety*:**
<font size=3>

O Kotlin foi desenhado para eliminar falhas de exceção de ponteiro nulo (`NullPointerException`). Como o projeto lida com leituras instáveis de sensores que podem retornar nulo (ex: perdas momentâneas de sinal Wi-Fi), entender o sistema de nulos é indispensável.

* **Tipos Anuláveis (`Type?`):** Por padrão, uma variável não aceita `null`. Para permitir valores nulos, adiciona-se o sufixo `?`.

* **Safe Call Operator (`?.`):** Executa o método apenas se o valor não for nulo; caso contrário, retorna `null`.

* **Elvis Operator (`?:`):** Fornece um valor alternativo seguro caso o lado esquerdo resulte em `null`.

````kotlin
val detectedBssid: String? = null

// Tentativa de leitura do tamanho da String com Safe Call e Elvis Operator
val bssidLength = detectedBssid?.length ?: 0
val formattedMessage = detectedBssid?.uppercase() ?: "ROTEADOR_DESCONHECIDO"

println("BSSID Formatado: $formattedMessage (Tamanho: $bssidLength)")
````

### **3. POO e Modelagem com `data class`:**
<font size=3>

Nesta etapa, iremos aprender a criar estruturas de dados leves para representar os objetos do mapa do prédio.

* **`data class`:** Em Kotlin, criar uma classe para armazenar dados exige apenas uma linha de código. O compilador gera automaticamente métodos de comparação (`equals`), cópia (`copy`) e exibição formatada (`toString`).

* **Enums:** Para representar direções e estados fixos da navegação.

````kotlin
enum class SensorType { WIFI, ACCELEROMETER, GYROSCOPE }

data class SensorReading(val deviceId: String,
                         val value: Double,
                         val type: SensorType,
                         val timestamp: Long = System.currentTimeMillis())

val reading1 = SensorReading("ESP32_01", -55.0, SensorType.WIFI)
println("Objeto criado: $reading1")
````

### **4. Programação Funcional e Operações em Coleções:**
<font size=3>

Uma das maiores vantagens do Kotlin para o projeto é a manipulação de listas e coleções usando programação funcional (`map`, `filter`, `sumOf`, `groupBy`). Isso facilitará a busca por roteadores e a navegação nos nós do grafo nas fases seguintes.

* **Listas Imutáveis (`listOf`) e Mutáveis (`mutableListOf`).**

* **`filter`:** Seleciona elementos que atendem a uma condição.

* **`map`:** Transforma cada elemento de uma lista em outro objeto.

* **`groupBy`:** Agrupa elementos por uma chave (ex: agrupar roteadores por andar/piso).

````kotlin
val readings = listOf(SensorReading("AP_01", -45.0, SensorType.WIFI),
                     SensorReading("AP_02", -80.0, SensorType.WIFI),
                     SensorReading("AP_03", -60.0, SensorType.WIFI))

// Filtrar apenas roteadores com sinal forte (maior que -70 dBm)
val strongSignals = readings
                    .filter { it.value > -70.0 }
                    .map { "Roteador ${it.deviceId} -> Sinal: ${it.value} dBm" }

println("Sinais adequados para Trilateração:")
strongSignals.forEach { println(it) }
````
