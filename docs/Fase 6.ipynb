### **2. Métricas de Precisão do IPS (Erro em Metros):**
<font size=3>

Para saber se o posicionamento é confiável, o grupo deve calcular o desvio entre a posição real onde o usuário está pisando $(x_{\text{real}}, y_{\text{real}})$ e a posição calculada pelo app $(x_{\text{est}}, y_{\text{est}})$.

* **Erro Euclidiano de Posicionamento ($e$):**
    $$e = \sqrt{(x_{\text{est}} - x_{\text{real}})^2 + (y_{\text{est}} - y_{\text{real}})^2}$$
    
* **Erro Médio Absoluto (MAE):** A média simples de todos os erros em metros durante um percurso de teste.

* **Erro no Percentil 90 ($P_{90}$):** Representa a margem de erro máxima esperada em $90\%$ do tempo (ex: "Em $90\%$ das vezes, o erro de localização é menor que 2,5 metros").
      

```kotlin
import kotlin.math.pow
import kotlin.math.sqrt
```

```kotlin
data class RealAndEstimatedPoint(val realX: Double, val realY: Double,
                                 val estimatedX: Double, val estimatedY: Double)

fun calculateMeanAbsoluteError(points: List<RealAndEstimatedPoint>): Double {
    val errors = points.map { p ->
        sqrt((p.estimatedX - p.realX).pow(2) + (p.estimatedY - p.realY).pow(2))
    }
    return errors.average()
}
```
